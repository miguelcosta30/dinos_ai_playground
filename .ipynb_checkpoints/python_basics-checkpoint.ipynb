#### 1.0 Basic concepts

##### 1.1 Types

```python
dinosaur_name = "Pintarolas"
dinosaur_species = "Triceratops"
dinosaur_age = 25
dinosaur_weight = 1325.25
dinosaur_parents= ["Pantufas","Barbatanas"]
is_extinct = True
last_seen = None


print(f"""
🦖 DINOSAUR PROFILE
----------------------------
Name        : {dinosaur_name}
Species     : {dinosaur_species}
Age         : {dinosaur_age} years
Weight      : {dinosaur_weight:.2f} kg
Parents     : {", ".join(dinosaur_parents)}
Extinct     : {"Yes" if is_extinct else "No"}
Last Seen   : {last_seen if last_seen else "Unknown"}
----------------------------
""")
```

##### 1.2 Dictionaries

```python
dinosaur = {
    "Name": dinosaur_name,
    "Species": dinosaur_species,
    "Age": f"{dinosaur_age} years",
    "Weight": f"{dinosaur_weight:.2f} kg",
    "Parents": ", ".join(dinosaur_parents),
    "Extinct": "Yes" if is_extinct else "No",
    "Last Seen": last_seen or "Unknown"
}

print("🦖 DINOSAUR PROFILE")
print("-" * 30)

for key, value in dinosaur.items():
    print(f"{key:<12}: {value}")
print("-" * 30)
```

##### 1.3 Classes

```python
class Dinosaur:
    def __init__(
        self,
        name: str,
        species:str ,
        weight: float,
        age:int,
        parents:list[str],
        extinct:bool,
        last_seen:str | None = None,
    ):
        self.name = name
        self.species = species
        self.age = age
        self.weight = weight
        self.parents = parents
        self.is_extinct = is_extinct
        self.last_seen = last_seen

    def __str__(self):
        return f"{self.name} ({self.species})"

dinosaur = Dinosaur("Pintaloras", "Triceratops",1243.24, 25, ["Pantufas","Barbatanas"], True)
print(dinosaur)
print(str(dinosaur.weight) + ' kg')
print(f"{dinosaur.weight} kg")
```
